```python
"""
Location Extractor for Power Outage Data
Extracts structured location information from raw text data
"""

import re
import json

# Raw data
raw_data = """
Bhikapada village under Kutunia panchayat and Geerjapada village under Malidihi panchayat of Rajgangpur block due to burning of transformer.
Kharamunda village Bijepur block. (Bargarh dist)
Karangamal village lift point under Boden block of Nuapada dist.
Srirampada, Po-Sahala, Sonepur dist.
Dharamgarh area citing increasing theft cases.(Kalahandi dist)
Nidhenpali village of Lupursingha GP under Bheden block, Bargarh dist.
Jhengnadihi village of Bhengarajpur GP under Paikmal. (Bargarh dist)
Tikrapada area, Bolangir. (Section-2)
Sarabhal, Jharsuguda
Thuamul, Rampur block Bhawnipatna Kahalandi
Nuagaon block  since 7:45 AM ,Dist sundargarh
chhend  Rourkela.
Murusundha, Lupursingha Fidder, Salebhata Section, Bolangir dist.
Hirakud area  since 12HRS.
Desil village, Titlagarh since yesterday 9 PM.
Jharmunda and Bijubandha slum of Tangarpali area, South Rourkela.
"""

def extract_locations(text):
    """
    Extract location information from the text
    Returns a list of dictionaries with structured location data
    """
    locations = []
    lines = text.strip().split('\n')
    
    for line in lines:
        if not line.strip():
            continue
            
        location_data = {
            'villages': [],
            'panchayat': None,
            'block': None,
            'district': None,
            'area': None,
            'additional_info': [],
            'original_text': line.strip()
        }
        
        # Extract district
        dist_match = re.search(r'(\w+)\s+dist', line, re.IGNORECASE)
        if dist_match:
            location_data['district'] = dist_match.group(1).strip()
        
        dist_match2 = re.search(r'Dist\s+(\w+)', line, re.IGNORECASE)
        if dist_match2:
            location_data['district'] = dist_match2.group(1).strip()
        
        # Extract block
        block_match = re.search(r'(\w+)\s+block', line, re.IGNORECASE)
        if block_match:
            location_data['block'] = block_match.group(1).strip()
        
        # Extract villages
        village_matches = re.findall(r'(\w+)\s+village', line, re.IGNORECASE)
        location_data['villages'] = [v.strip() for v in village_matches]
        
        # Extract panchayat
        panchayat_match = re.search(r'(\w+)\s+panchayat', line, re.IGNORECASE)
        if panchayat_match:
            location_data['panchayat'] = panchayat_match.group(1).strip()
        
        # Extract GP (Gram Panchayat)
        gp_match = re.search(r'(\w+)\s+GP', line, re.IGNORECASE)
        if gp_match:
            location_data['panchayat'] = gp_match.group(1).strip()
        
        # Extract area
        area_match = re.search(r'(\w+)\s+area', line, re.IGNORECASE)
        if area_match:
            location_data['area'] = area_match.group(1).strip()
        
        # Extract additional location names (before commas or specific keywords)
        # This captures places like "Sarabhal", "Chhend", etc.
        simple_location = re.match(r'^([A-Za-z\s]+),', line)
        if simple_location and not location_data['villages']:
            location_data['area'] = simple_location.group(1).strip()
        
        # Extract additional info like "slum", "lift point", etc.
        if 'slum' in line.lower():
            location_data['additional_info'].append('slum')
        if 'lift point' in line.lower():
            location_data['additional_info'].append('lift point')
        
        locations.append(location_data)
    
    return locations

def print_locations_summary(locations):
    """Print a clean summary of all locations"""
    print("=" * 80)
    print("EXTRACTED LOCATIONS SUMMARY")
    print("=" * 80)
    
    all_places = set()
    
    for i, loc in enumerate(locations, 1):
        print(f"\n{i}. ", end="")
        
        location_parts = []
        
        # Villages
        if loc['villages']:
            for village in loc['villages']:
                all_places.add(village)
            location_parts.append(f"Villages: {', '.join(loc['villages'])}")
        
        # Area
        if loc['area']:
            all_places.add(loc['area'])
            location_parts.append(f"Area: {loc['area']}")
        
        # Panchayat
        if loc['panchayat']:
            all_places.add(loc['panchayat'])
            location_parts.append(f"Panchayat: {loc['panchayat']}")
        
        # Block
        if loc['block']:
            all_places.add(loc['block'])
            location_parts.append(f"Block: {loc['block']}")
        
        # District
        if loc['district']:
            all_places.add(loc['district'])
            location_parts.append(f"District: {loc['district']}")
        
        # Additional info
        if loc['additional_info']:
            location_parts.append(f"Type: {', '.join(loc['additional_info'])}")
        
        print(" | ".join(location_parts))
    
    print("\n" + "=" * 80)
    print(f"UNIQUE PLACES FOUND: {len(all_places)}")
    print("=" * 80)
    print(", ".join(sorted(all_places)))
    print()

def export_to_csv(locations, filename='locations.csv'):
    """Export locations to CSV file"""
    import csv
    
    with open(filename, 'w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        writer.writerow(['Villages', 'Panchayat', 'Block', 'District', 'Area', 'Additional Info', 'Original Text'])
        
        for loc in locations:
            writer.writerow([
                ', '.join(loc['villages']) if loc['villages'] else '',
                loc['panchayat'] or '',
                loc['block'] or '',
                loc['district'] or '',
                loc['area'] or '',
                ', '.join(loc['additional_info']) if loc['additional_info'] else '',
                loc['original_text']
            ])
    
    print(f"✓ Data exported to {filename}")

def export_to_json(locations, filename='locations.json'):
    """Export locations to JSON file"""
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(locations, f, indent=2, ensure_ascii=False)
    
    print(f"✓ Data exported to {filename}")

# Main execution
if __name__ == "__main__":
    # Extract locations
    locations = extract_locations(raw_data)
    
    # Print summary
    print_locations_summary(locations)
    
    # Export to files
    export_to_csv(locations, 'locations.csv')
    
    print("\n" + "=" * 80)
    print("EXTRACTION COMPLETE!")
    print("=" * 80)
```

```python
"""
Location Extractor for Power Outage Data
Extracts structured location information from raw text data
"""

import re
import json

# Raw data
raw_data = """
Bhikapada village under Kutunia panchayat and Geerjapada village under Malidihi panchayat of Rajgangpur block due to burning of transformer.
Kharamunda village Bijepur block. (Bargarh dist)
Karangamal village lift point under Boden block of Nuapada dist.
Srirampada, Po-Sahala, Sonepur dist.
Dharamgarh area citing increasing theft cases.(Kalahandi dist)
Nidhenpali village of Lupursingha GP under Bheden block, Bargarh dist.
Jhengnadihi village of Bhengarajpur GP under Paikmal. (Bargarh dist)
Tikrapada area, Bolangir. (Section-2)
Sarabhal, Jharsuguda
Thuamul, Rampur block Bhawnipatna Kahalandi
Nuagaon block  since 7:45 AM ,Dist sundargarh
chhend  Rourkela.
Murusundha, Lupursingha Fidder, Salebhata Section, Bolangir dist.
Hirakud area  since 12HRS.
Desil village, Titlagarh since yesterday 9 PM.
Jharmunda and Bijubandha slum of Tangarpali area, South Rourkela.
"""

def extract_locations(text):
    """
    Extract all location names from the text
    Returns a list of location strings in simple format
    """
    location_lines = []
    lines = text.strip().split('\n')
    
    for line in lines:
        if not line.strip():
            continue
        
        locations = []
        
        # Extract villages (word before "village")
        village_matches = re.findall(r'(\w+)\s+village', line, re.IGNORECASE)
        locations.extend([v.strip() for v in village_matches])
        
        # Extract panchayats (word before "panchayat" or "GP")
        panchayat_matches = re.findall(r'(\w+)\s+(?:panchayat|GP)', line, re.IGNORECASE)
        locations.extend([p.strip() for p in panchayat_matches])
        
        # Extract blocks (word before "block")
        block_matches = re.findall(r'(\w+)\s+block', line, re.IGNORECASE)
        locations.extend([b.strip() for b in block_matches])
        
        # Extract districts (word before "dist" or after "Dist")
        dist_matches = re.findall(r'(?:(\w+)\s+dist|Dist\s+(\w+))', line, re.IGNORECASE)
        for match in dist_matches:
            dist_name = match[0] if match[0] else match[1]
            if dist_name:
                locations.append(dist_name.strip())
        
        # Extract areas (word before "area")
        area_matches = re.findall(r'(\w+)\s+area', line, re.IGNORECASE)
        locations.extend([a.strip() for a in area_matches])
        
        # Extract location names that appear before commas at start of line
        start_match = re.match(r'^([A-Za-z]+)\s*,', line)
        if start_match:
            loc_name = start_match.group(1).strip()
            if loc_name.lower() != 'po':  # Skip "Po-" prefix
                locations.append(loc_name)
        
        # Extract "word1 and word2" patterns (like "Jharmunda and Bijubandha")
        and_matches = re.findall(r'(\w+)\s+and\s+(\w+)', line, re.IGNORECASE)
        for match in and_matches:
            # Skip common words
            if match[0].lower() not in ['slum', 'of', 'under', 'since']:
                locations.append(match[0].strip())
            if match[1].lower() not in ['slum', 'of', 'under', 'since']:
                locations.append(match[1].strip())
        
        # Extract location after "Po-"
        po_match = re.search(r'Po-(\w+)', line, re.IGNORECASE)
        if po_match:
            locations.append(po_match.group(1).strip())
        
        # Extract city names (specific known cities in the region)
        city_patterns = ['Rourkela', 'Bhawanipatna', 'Bhawnipatna', 'Titlagarh', 'Jharsuguda', 'Bolangir']
        for city in city_patterns:
            if city.lower() in line.lower() and city not in locations:
                locations.append(city)
        
        # Remove duplicates and unwanted words while preserving order
        seen = set()
        unique_locations = []
        unwanted = {'panchayat', 'village', 'block', 'dist', 'gp', 'area', 'slum', 'of', 'under', 'po', 'since', 'section'}
        
        for loc in locations:
            if loc.lower() not in seen and loc.lower() not in unwanted:
                seen.add(loc.lower())
                unique_locations.append(loc)
        
        if unique_locations:
            location_lines.append(', '.join(unique_locations) + '.')
    
    return location_lines

def print_locations(locations):
    """Print locations in simple format"""
    print("=" * 80)
    print("EXTRACTED LOCATIONS")
    print("=" * 80)
    print()
    
    for loc_line in locations:
        print(loc_line)
    
    print()
    print("=" * 80)

def save_to_file(locations, filename='locations.txt'):
    """Save locations to text file"""
    with open(filename, 'w', encoding='utf-8') as f:
        for loc_line in locations:
            f.write(loc_line + '\n')
    
    print(f"✓ Locations saved to {filename}")

# Main execution
if __name__ == "__main__":
    # Extract locations
    locations_exact_df = extract_locations(raw_data)
    
    # Print locations
    print_locations(locations_exact_df)
    
    # Save to file
    save_to_file(locations_exact_df, 'locations_exact_df.txt')
    
    print("\n" + "=" * 80)
    print("EXTRACTION COMPLETE!")
    print("=" * 80)
```
