```python
import pandas as pd

# Define a function to classify values with inclusive boundaries
def classify_duration(x):
    if x <= 2:
        return "<2"
    elif 2 < x <= 4:
        return "2<4"
    elif 4 < x <= 8:
        return "4<8"
    elif x >= 8:
        return ">8"
    else:
        return None

# Apply function to create new column
close_open_hour["DURATION_RANGE"] = close_open_hour["DURATION_HOURS_INT"].apply(classify_duration)




```
