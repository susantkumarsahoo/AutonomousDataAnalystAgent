```python
import pandas as pd

def generate_month_open_clode_pivot_report(dataset_path: str) -> dict:
    # Load dataset
    new_df = pd.read_excel(dataset_path)

    # Clean and format columns
    new_df['DATE'] = pd.to_datetime(new_df['DATE'])
    df = new_df[new_df['DATE'].dt.to_period('M') == '2025-11']
    df['COMPLAINT TYPE'] = df['COMPLAINT TYPE'].astype(str).str.strip().str.title()
    df['DEPT'] = df['DEPT'].astype(str).str.strip().str.title()
    df['CLOSED/OPEN'] = df['CLOSED/OPEN'].astype(str).str.strip().str.title()

    pivot = pd.pivot_table(
        df,
        values='DATE',
        index=['COMPLAINT TYPE'],          # keep this index
        columns=['DEPT','CLOSED/OPEN'],
        aggfunc='count',
        fill_value=0,
        margins=True,
        margins_name='Grand Total',
        observed=False
    )

    # Flatten MultiIndex columns into single strings
    pivot.columns = [f"{dept}_{status}" for dept, status in pivot.columns]

    # Convert pivot table to dictionary format, preserving index
    pivot_dict = pivot.to_dict()

    return pivot_dict

```

```python
import pandas as pd

def generate_report(dataset_path: str) -> dict:
    # Load dataset
    df = pd.read_excel(dataset_path)

    # Ensure DATE is datetime
    df['DATE'] = pd.to_datetime(df['DATE'], errors='coerce')

    # Clean string columns
    string_columns = ['COMPLAINT TYPE', 'DEPT', 'CLOSED/OPEN']
    for col in string_columns:
        if col in df.columns:
            df[col] = df[col].astype(str).str.strip().str.title()

    # Helper function to safely get value counts
    def safe_value_counts(column_name):
        if column_name in df.columns:
            if column_name == 'DATE':
                # Convert dates to strings for serialization
                return df[column_name].dt.strftime('%Y-%m-%d').value_counts().to_dict()
            else:
                value_counts = df[column_name].value_counts()
                return {str(k): int(v) for k, v in value_counts.items()}
        return {}

    # Build report dictionary
    report = {
        "COMPLAINT TYPE": safe_value_counts('COMPLAINT TYPE'),
        "DATE": safe_value_counts('DATE'),
        "SHIFT DUTY": safe_value_counts('SHIFT DUTY'),
        "QUERY/REQUEST/COMPLAINT": safe_value_counts('QUERY/REQUEST/COMPLAINT'),
        "SECTION": safe_value_counts('SECTION'),
        "SUB-DIVISION": safe_value_counts('SUB-DIVISION'),
        "CIRCLE": safe_value_counts('CIRCLE'),
        "CONSUMER NUMBER": safe_value_counts('CONSUMER NUMBER'),
        "MOBILE NUMB": safe_value_counts('MOBILE NUMB'),
        "DEPT": safe_value_counts('DEPT'),
        "CLOSED/OPEN": safe_value_counts('CLOSED/OPEN'),
        "COMPLAINANT NAME": safe_value_counts('COMPLAINANT NAME'),
        "PSCC/FG/TO": safe_value_counts('PSCC/FG/TO'),
        "MINUTE": safe_value_counts('MINUTE')
    }

    return report
```

```python
import pandas as pd
from typing import Optional, Dict, Any, List

def generate_report(
    dataset_path: str,
    filters: Optional[Dict[str, Any]] = None,
    date_range: Optional[tuple] = None,
    columns: Optional[List[str]] = None
) -> dict:
    """
    Generate a report from Excel dataset with filtering options.
    
    Args:
        dataset_path: Path to the Excel file
        filters: Dictionary of column filters e.g., {'DEPT': 'Engineering', 'CLOSED/OPEN': 'Open'}
        date_range: Tuple of (start_date, end_date) as strings 'YYYY-MM-DD'
        columns: List of specific columns to include in report (None = all columns)
    
    Returns:
        Dictionary with value counts for each column
    """
    # Load dataset
    df = pd.read_excel(dataset_path)

    # Ensure DATE is datetime
    df['DATE'] = pd.to_datetime(df['DATE'], errors='coerce')

    # Clean string columns
    string_columns = ['COMPLAINT TYPE', 'DEPT', 'CLOSED/OPEN', 'COMPLAINANT NAME']
    for col in string_columns:
        if col in df.columns:
            df[col] = df[col].astype(str).str.strip().str.title()

    # Apply filters
    if filters:
        for column, value in filters.items():
            if column in df.columns:
                if isinstance(value, list):
                    # Multiple values (OR condition)
                    df = df[df[column].isin(value)]
                else:
                    # Single value
                    df = df[df[column] == value]

    # Apply date range filter
    if date_range:
        start_date, end_date = date_range
        if start_date:
            df = df[df['DATE'] >= pd.to_datetime(start_date)]
        if end_date:
            df = df[df['DATE'] <= pd.to_datetime(end_date)]

    # Determine which columns to include
    if columns is None:
        columns = [
            "COMPLAINT TYPE", "DATE", "SHIFT DUTY", "QUERY/REQUEST/COMPLAINT",
            "SECTION", "SUB-DIVISION", "CIRCLE", "CONSUMER NUMBER",
            "MOBILE NUMB", "DEPT", "CLOSED/OPEN", "COMPLAINANT NAME",
            "PSCC/FG/TO", "MINUTE"
        ]

    # Helper function to safely get value counts
    def safe_value_counts(column_name):
        if column_name in df.columns:
            if column_name == 'DATE':
                # Convert dates to strings for serialization
                return df[column_name].dt.strftime('%Y-%m-%d').value_counts().to_dict()
            else:
                value_counts = df[column_name].value_counts()
                return {str(k): int(v) for k, v in value_counts.items()}
        return {}

    # Build report dictionary
    report = {
        "total_records": len(df),
        "filters_applied": filters if filters else {},
        "date_range": date_range if date_range else None,
        "data": {}
    }

    for col in columns:
        if col in df.columns:
            report["data"][col] = safe_value_counts(col)

    return report


```
